```python
#10. What are the ACID properties in the context of MongoDB transactions?
# ACID is an acronym that stands for:

# Atomicity: This means that all operations within a transaction are treated as a single, indivisible unit of work. Either all operations succeed, or none of them do. There's no "in-between" state.
# Consistency: A transaction must maintain the database's integrity constraints. It transforms the database from one valid state to another. If any rule or constraint is violated during a transaction, the transaction is rolled back.
# Isolation: Transactions are isolated from each other. Concurrent transactions do not interfere with each other's operations. The effect is as if transactions were executed one at a time, even if they actually run concurrently.
# Durability: Once a transaction is committed, the changes are permanent, even in the event of system failures like power outages or crashes. The database guarantees that the data will survive such events.
```

```python
#12. How does MongoDB handle schema validation?
# MongoDB provides schema validation to enforce data quality and consistency within your collections.  It allows you to specify rules that documents must adhere to before they can be inserted or updated.  Here's a breakdown of how it works:

# 1. Defining Schemas:

# You define schemas using JSON Schema.  This schema specifies the structure and data types of the documents you expect to store in a collection.  You can define:

# Required fields: Fields that must be present in every document.
# Data types: The allowed data types for each field (e.g., string, number, boolean, array, object).
# Value constraints: Restrictions on the values of fields, such as minimum/maximum values for numbers, regular expressions for strings, or allowed values in an array.
# Document structure: The nesting and arrangement of fields within the document.
# 2. Applying Schemas:

# You can apply a schema to a collection using the db.createCollection() command (when creating a new collection) or the db.runCommand() command with the collMod command (to modify an existing collection).
# 3. Validation Enforcement:

# Once a schema is applied, MongoDB will enforce it whenever you attempt to insert or update a document in the collection.

# Insertion: If a new document doesn't conform to the schema, the insertion will fail, and MongoDB will return an error message.
# Update: Similarly, if an update operation would cause a document to violate the schema, the update will fail.
# 4. Validation Levels:

# MongoDB provides different validation levels to control how strictly the schema is enforced:

# off: Schema validation is disabled.
# moderate: MongoDB validates documents only during inserts and updates. Existing documents that violate the schema are not affected. This is the default.
# strict: MongoDB validates all documents, including existing ones. If you enable strict validation on a collection with existing documents that violate the schema, it will result in an error.
# You can set the validation level when creating or modifying a collection.

# 5. Validation Action:

# You can also specify the validationAction:

# error: The default. If a document violates the schema, the operation fails and an error is returned.
# warn: If a document violates the schema, the operation succeeds, but MongoDB logs a warning message. This can be useful for gradually enforcing a schema on existing data.
# 6. Benefits of Schema Validation:

# Data quality: Ensures that only valid data is stored in your database.
# Data consistency: Helps maintain consistency across your data by enforcing a uniform structure.
# Application reliability: Reduces the risk of application errors caused by unexpected data formats.
# Simplified development: Makes it easier for developers to work with data, as they can rely on the schema to define the expected structure.
# 7. Considerations:

# Performance: Schema validation can add some overhead to write operations. However, the benefits of data quality often outweigh this cost.
# Schema design: Carefully design your schemas to balance flexibility and strictness. Avoid overly restrictive schemas that may make it difficult to evolve your data model in the future.
```

```python
#13. What is the difference between a primary and a secondary node in a replica set?
# In a MongoDB replica set, the primary and secondary nodes play distinct roles in ensuring high availability and data consistency. Here's a breakdown of their differences:

# Primary Node:

# Handles all write operations: The primary node is the only member of the replica set that accepts write requests (inserts, updates, deletes). All client write operations are directed to the primary.
# Maintains the oplog: The primary node maintains the oplog (operations log), which is a record of all write operations performed on the primary. This oplog is crucial for the secondaries to replicate data.
# Default read operations: By default, client read operations are also directed to the primary, although you can configure your application to read from secondaries for read scaling.
# Secondary Nodes:

# Replicate data from the primary: Secondary nodes continuously replicate data from the primary's oplog. They apply the operations recorded in the oplog to their own datasets, ensuring they stay consistent with the primary.
# Read operations (optional): Secondaries can be used for read operations. This can help distribute the read load and improve performance, especially for read-heavy applications. However, reads from secondaries might not always be the most up-to-date (eventual consistency).
# Automatic failover: If the primary node fails, the secondaries are eligible to be elected as the new primary. This automatic failover process ensures high availability.
# Backup and recovery: Secondaries can be used for backups, as they maintain a copy of the data.
```

```python
#18. What are the advantages of using MongoDB for horizontal scaling?
# MongoDB's architecture lends itself well to horizontal scaling, offering several key advantages:

# Sharding: This is MongoDB's primary method for horizontal scaling.  Sharding distributes data across multiple servers (shards) based on a shard key.  This allows you to:

# Increase write capacity: Writes are distributed across shards, increasing the overall write throughput.
# Increase read capacity: Reads can be directed to the relevant shard(s), improving read performance.
# Increase storage capacity: As your data grows, you can add more shards to the cluster, effectively increasing your storage capacity.
# Automatic Data Distribution: MongoDB handles the distribution of data across shards automatically.  You define the shard key, and MongoDB takes care of splitting the data and placing it on the appropriate shards.  This simplifies the management of a distributed database.

# Scalability on Demand:  You can add or remove shards as needed to scale your database up or down based on your application's requirements.  This flexibility allows you to adapt to changing workloads and avoid over-provisioning resources.

# High Availability:  Sharding can also improve high availability.  If one shard goes down, the other shards can continue to serve requests.  Combined with replica sets (which provide redundancy within each shard), sharded clusters offer a robust and highly available solution.

# Improved Query Performance:  When queries are targeted using the shard key, they can be routed directly to the shard containing the relevant data. This significantly reduces the amount of data that needs to be scanned, resulting in faster query performance.

# Simplified Operations: While managing a sharded cluster is more complex than managing a single server, MongoDB provides tools and features that simplify operations.  For example, the mongos router acts as a single point of entry for applications and handles routing requests to the appropriate shards.

# Cost-Effectiveness: Horizontal scaling can be more cost-effective than vertical scaling (increasing the resources of a single server).  You can use commodity hardware for your shards, rather than relying on expensive, high-end servers.

# Fault Isolation: If one shard experiences a problem, it is less likely to affect the other shards in the cluster. This fault isolation can limit the impact of failures and improve the overall stability of your database.
```

```python
# 19. How do MongoDB transactions differ from SQL transactions?
# While both MongoDB and SQL databases offer transactions to ensure data consistency, there are some key differences in how they are implemented and the guarantees they provide:

# 1. Scope:

# SQL Transactions: Typically operate across multiple tables within a single database. They are designed to maintain relationships and consistency across related data.
# MongoDB Transactions: Initially focused on single-document atomicity. However, with the introduction of multi-document transactions in version 4.0, MongoDB now supports transactions that span multiple documents within a single collection or across multiple collections within the same database.
# 2. ACID Properties:

# SQL Transactions: Traditionally adhere strictly to all four ACID properties:
# Atomicity: All changes within a transaction are treated as a single, indivisible unit of work. Either all changes are made, or none are.
# Consistency: A transaction maintains the database's integrity constraints, ensuring that the database remains in a valid state before and after the transaction.
# Isolation: Transactions are isolated from each other, meaning that concurrent transactions do not interfere with each other's operations.
# Durability: Once a transaction is committed, the changes are persistent and survive even system failures.
# MongoDB Transactions: Also provide ACID guarantees, but with some nuances:
# Atomicity: MongoDB ensures atomicity for both single-document and multi-document transactions.
# Consistency: MongoDB maintains data consistency within a transaction, but the level of consistency might vary depending on the read concern and write concern configurations used.
# Isolation: MongoDB offers different isolation levels, allowing you to balance consistency and performance. By default, MongoDB provides snapshot isolation.
# Durability: MongoDB ensures durability through write concerns and replica sets, ensuring that committed transactions are written to a majority of replica set members.
#
# 3. Implementation:

# SQL Transactions: Typically involve locking mechanisms to ensure isolation. When a transaction accesses data, it might acquire locks to prevent other transactions from modifying that data until the transaction is complete.
# MongoDB Transactions: Utilize a different approach, relying on a distributed transaction coordinator and logical clocks to manage transactions. MongoDB uses optimistic concurrency control, which generally avoids long-lasting locks and can improve performance.
# 4. Data Model:

# SQL Transactions: Work within the context of a relational data model, where data is structured in tables with predefined schemas.
# MongoDB Transactions: Operate on a document-oriented data model, where data is stored in flexible, schema-less documents. This difference in data models can influence how transactions are designed and used.
# 5. Use Cases:

# SQL Transactions: Are well-suited for applications that require strong consistency and data integrity, such as financial systems, e-commerce platforms, and applications with complex relationships between data.
# MongoDB Transactions: Are suitable for applications that need to maintain data consistency across multiple operations, even in the context of a flexible data model.

#  They are often used in scenarios involving e-commerce, content management, and other applications where data consistency is important but the data model might be more dynamic.
```

```python
# 20. What are the main differences between capped collections and regular collections?
# Capped collections and regular collections in MongoDB serve different purposes and have distinct characteristics. Here's a breakdown of their main differences:

# 1. Size and Structure:

# Capped Collections:
# Fixed size: You define the maximum size (in bytes) or the maximum number of documents when creating a capped collection.
# Circular buffer: Once the collection reaches its limit, the oldest documents are automatically overwritten by new ones, ensuring a fixed size.
# Insertion order maintained: Documents are stored and retrieved in the order they were inserted.
# Regular Collections:
# Dynamic size: Can grow or shrink as needed.
# No automatic deletion: Documents are not automatically removed unless explicitly deleted or expired using features like TTL indexes.
# No guaranteed order: Unless explicitly sorted, the order of documents retrieved from a regular collection is not guaranteed.
# 2. Performance:

# Capped Collections:
# High insertion throughput: Maintaining insertion order without requiring indexes improves write performance.
# Efficient for specific use cases: Ideal for situations where you only need the most recent data (e.g., logs, sensor readings).
# Regular Collections:
# Flexible: Can be optimized for various query patterns by creating indexes.
# More general-purpose: Suitable for most data storage needs where data might need to be queried in different ways.
# 3. Restrictions:

# Capped Collections:
# Cannot be sharded: Data in a capped collection cannot be distributed across multiple shards.
# Limited update operations: Updating documents in a capped collection can be tricky as it might cause the document to grow in size, potentially violating the fixed-size constraint.
# No manual deletion: You can't delete individual documents from a capped collection; you can only remove all documents or let the oldest ones be overwritten.
# Regular Collections:
# More flexible: Can be sharded, indexed, and manipulated in various ways.
# 4. Use Cases:

# Capped Collections:
# Log storage: Keeping recent log entries for debugging or monitoring.
# Caching: Storing frequently accessed data that can be discarded when it's no longer needed.
# Real-time data streams: Handling sensor data or stock tickers where only the latest information is relevant.
# Regular Collections:
# General data storage: Suitable for most applications where you need to store and query data in a flexible way.
# User data: Storing user profiles, preferences, and other information.
# Product catalogs: Managing product details, categories, and inventory.
# In essence:

# Capped collections are designed for specific use cases where you need a fixed-size collection with high insertion throughput and automatic removal of old data. They are like a circular buffer that always keeps the most recent data.
# Regular collections are more versatile and suitable for general-purpose data storage where you need flexibility in managing and querying your data.
```

```python
# 23. What is MongoDB’s WiredTiger storage engine, and why is it important?
# WiredTiger is the default storage engine for MongoDB since version 3.2. It's a crucial component because it's responsible for how your data is stored, retrieved, and managed on disk. Think of it as the engine that powers MongoDB's data handling.

# Here's why WiredTiger is so important:

# 1. Performance:

# Document-level Concurrency: WiredTiger allows multiple write operations to occur simultaneously on different documents within the same collection. This significantly improves write throughput and reduces contention, making MongoDB more efficient for applications with high write loads.
# Optimistic Concurrency Control: WiredTiger generally avoids long-lasting locks, which can hinder performance. It uses an optimistic approach, assuming that conflicts between operations are rare. If a conflict does occur, it transparently retries the operation.
# Cache Management: WiredTiger uses an internal cache to manage data in memory, optimizing read and write performance. It efficiently manages memory usage to ensure fast access to frequently used data.
# 2. Data Integrity and Durability:

# ACID Properties: WiredTiger supports ACID properties, ensuring that transactions are atomic, consistent, isolated, and durable. This is crucial for maintaining data integrity, especially in applications where data consistency is paramount.
# Checkpointing: WiredTiger periodically writes snapshots of the data to disk, creating consistent recovery points. This ensures data durability and enables fast recovery in case of failures.
# Journaling: WiredTiger uses a write-ahead log (journal) to record all data modifications between checkpoints. This ensures that even if MongoDB crashes, changes made since the last checkpoint can be recovered from the journal.
# 3. Storage Efficiency:

# Data Compression: WiredTiger supports compression for both collections and indexes, reducing storage space requirements and I/O operations. This can lead to significant cost savings and improved performance.
# Variable-Length Pages: WiredTiger uses variable-length pages, which means there's less wasted space for large objects. This also eliminates the need for compaction as pages grow and shrink naturally when data is inserted or deleted.
# 4. Scalability:

# Scalable Architecture: WiredTiger is designed to scale on modern multi-core architectures. It can efficiently handle large datasets and high traffic loads, making MongoDB suitable for demanding applications.
```

```python
#1. Write a Python script to load the Superstore dataset from a CSV file into MongoDB.
import pandas as pd
from pymongo import MongoClient

def load_superstore_to_mongodb(csv_file_path, mongodb_uri, database_name, collection_name):
    """Loads the Superstore dataset from a CSV file into MongoDB.

    Args:
        csv_file_path: Path to the Superstore CSV file.
        mongodb_uri: MongoDB connection URI (e.g., "mongodb://localhost:27017/").
        database_name: Name of the MongoDB database.
        collection_name: Name of the MongoDB collection.
    """

    try:
        # 1. Read the CSV file using pandas
        df = pd.read_csv(csv_file_path, encoding='unicode_escape') #Handles encoding issues

        # 2. Convert the DataFrame to a list of dictionaries (required for MongoDB)
        data = df.to_dict(orient='records')

        # 3. Connect to MongoDB
        client = MongoClient(mongodb_uri)
        db = client[database_name]
        collection = db[collection_name]

        # 4. Insert the data into the collection
        result = collection.insert_many(data)

        print(f"Successfully inserted {len(result.inserted_ids)} documents into {collection_name}")

    except FileNotFoundError:
        print(f"Error: CSV file not found at {csv_file_path}")
    except Exception as e:
        print(f"An error occurred: {e}")
    finally:
        if 'client' in locals() and client:  # Ensure client is defined before closing
            client.close()


# Example usage:  (REPLACE WITH YOUR ACTUAL DETAILS)
csv_file = "Superstore.csv"  # Replace with the actual path to your CSV file
mongo_uri = "mongodb://localhost:27017/"  # Replace with your MongoDB URI
db_name = "superstore_db"  # Replace with your desired database name
collection_name = "superstore_data"  # Replace with your desired collection name

load_superstore_to_mongodb(csv_file, mongo_uri, db_name, collection_name)
```
